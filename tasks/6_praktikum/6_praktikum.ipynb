## 4. Aufgabe: Labyrinth

In dieser Aufgabe soll der Algorithmus zum Verlassen eines Labyrinths aus dem [Einleitungskapitel](https://fb03_laborti.fh-muenster.io/grundlagen_der_programmierung/skriptum/einfuehrung.html) des Skriptums umgesetzt werden.

Das Labyrinth ist in Form der monochromen Bilddatei `labyrinth.tiff` gegeben. Es kommen zwei Graustufen vor, mit folgenden Bedeutungen:

- Wert $=0$: Wand
- Wert $=255$: Unbesuchter Weg

Eine Person, die sich durch das Labyrinth bewegt, soll programmiertechnisch durch ein Dictionary dargestellt werden, welches folgende Attribute hat:
- `map`: Karte des Labyrinths, welche die gegangenen Wege sowie die erkundeten Wände enthält.
    - Wert $=0$: Wand
    - Wert $=255$: Unerkundetes Gebiet
    - $0 <$ Wert $<255$: Besuchter Weg
- `position`: Position in kartesischen Koordinaten innerhalb des Labyrinths. Die Logik hinter den Koordinaten entspricht der Zeilen- und Spaltenindizierung einer Matrix. Je größer die Zeilenkoordinate (Zeilenindex), desto weiter *unten* im Labyrinth befindet sich die Person. Je größer die Spaltenkoordinate (Spaltenindex), desto weiter *rechts* im Labyrinth befindet sich die Person.
- `orientation`: Orientierung (Blickrichtung) der Person.
    - 0: hoch
    - 1: rechts
    - 2: unten
    - 3: links

In den Teilaufgaben werden wiederkehrende Programmteile zunächst als Funktion definiert, die im letzten Aufgabenteil zu einem lauffähigen Algorithmus zusammengesetzt werden.

Vorab sei hier eine Übersicht der zu implementierenden Funktionen gegeben:
- `init_person`: Initialisiert das Dictionary.
- `get_direction`: Richtungsvektor gemäß der Orientierung.

- `is_step_possible`: Erkennt Wände und gibt zurück, ob ein Schritt in diese Richtung möglich ist.
- `go_one_step`: Geht einen Schritt in die aktuelle Blickrichtung.
- `turn_right`: Person dreht sich 90° rechts herum.
- `turn_left`: Person dreht sich 90° links herum.
- `is_at_end_position`: Gibt zurück, ob der Ausgang gefunden wurde.
